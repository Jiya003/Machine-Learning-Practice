```python
#defining X and Y
X=df['Hours']
Y=df['Scores']
```

```python
import statsmodels.api as sm

X=sm.add_constant(X)
```

```python
model=sm.OLS(Y,X).fit()
```

```python
print(model.summary())
```

```python
Y_pred=model.predict(X)
```

```python
import matplotlib.pyplot as plt

#We are using df['Hours'] because X has an intercept
plt.scatter(df['Hours'],Y,color="blue")
plt.plot(df['Hours'],Y_pred,color="red")
plt.show()
```
