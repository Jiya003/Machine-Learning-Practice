Predict House Prices

    Dataset: house_data.csv with features like Area, Bedrooms, Location, Price

    Task:

        Build an SLR model to predict Price based on Area

        Train/test split

        Predict on test data

        Save predictions to house_price_predictions.csv

```python
import pandas as pd
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

X=df[['area']]
Y=df['price']

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)

print(type(Y_train),Y_train.shape)
print(type(X_train), X_train.shape)

model=LinearRegression()

model.fit(X_train,Y_train)

Y_pred=model.predict(X_test)
```

```python
import numpy as np

mae=mean_absolute_error(Y_test,Y_pred)
mse=mean_squared_error(Y_test,Y_pred)
rmse=np.sqrt(mse)
r2=r2_score(Y_test,Y_pred)
```

```python
print(f"The mean absolute error is: {mae:.2f}")
print(f"The mean squared error is: {mse:.2f}")
print(f"The root mean squared error is: {rmse:.2f}")
print(f"R2 score is: {r2:.2f}")
```
